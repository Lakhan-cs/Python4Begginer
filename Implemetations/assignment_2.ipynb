#### Calculate area of a triangle

```python
a = float(input('Enter first side : '))
b = float(input('Enter second side : '))
c = float(input('Enter third side : '))

s = (a + b + c) / 2

area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print('The area of the triangle is %0.2f' % area)
```

#### fibonacci series

```python
n = int(input("Enter a number : "))
a, b = 0, 1
count = 0 

print(f"Fabonacci series : ", end="")
while count<n: 
    a, b = b, a+b
    count += 1
    print(a, end=" ")
```

#### Input numbers gets negative value and then print sum of all inputs

```python
total = 0
while True:
    num=float(input("Enter a non - negative number : "))
    if num<0:
        break
    total += num

print(f"The sum of inputs is {total}")
```

#### `5x5 grid of *` 

```python
for i in range(5):
    for j in range(5):
        print("*", end=" ")
    print()
```

#### Find sum of digits of a number

```python
num = int(input("Enter a num : "))
sum = 0

while num > 0 :

    lastdigit = num % 10
    
    sum += lastdigit

    num = num // 10

print(f"The sum of digits of {num} is : {sum}")
```

#### prime number check

```python
num=int(input("Enter a num : "))

is_prime = True 
if num<=1:
    is_prime=False
else : 
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            is_prime=False
            break

if is_prime:
    print(f"{num} is a prime number.")
else :
    print(f"{num} is not a prime number.")
```
