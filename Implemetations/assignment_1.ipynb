#### Check if number is positive / negative / zero

```python
a = int(input("Enter a num :"))

if a>0:
    print(a,"is positive")
elif a==0:
    print(a,"is zero")
elif a<0:
    print(a,"is negative")
```

#### Find largest from three numbers

```python
a = float(input("first number :"))
b = float(input("second number :"))
c = float(input("third number :"))


if (a>b) and (a>c):
    print("largest :", a)
elif (b>a) and (b>c):
    print("largest :", b)
elif (c>a) and (c>b):
    print("largest :", c)
```

#### Factorial of a number using recursion

```python
def factorial(n):
    if n == 0:
        return 1
    else :
        return n*factorial(n-1)
    
num = int(input("Enter a num : "))
print(f"The factorial of {num} is {factorial(num)}")
```

#### Swaping two variable 

```python
a = 10 
b = 20 

print(f"before : a = {a}, b = {b}")  # before swap


a, b = b, a 
print(f"after : a = {a}, b = {b}") # after swap
```

#### Find the sum of first n natural numbers

```python
n = int(input("Enter a num : "))

sum_n = (n * (n+1))//2 

print(f"The sum of first {n} natural num is {sum_n}")
```

#### Find if a year is leap year or not

```python
year = int(input("Enter a year : "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")

else : 
    print(f"{year} is a normal year.")
```

#### palindrome check in strings

```python
string = input("Enter a str : ")
rev_str = string[::-1]

print(f"The reversed of {string} is {rev_str}")



if string == string[::-1]:
    print(f"{string} is palindrome.")
else : 
    print(f"{string} is not palindrome.")
```
