#### keyword args  - parameters with dict format

```python
def books(title, author, genre):

    bookCollection = {'title':title, "author":author}
    print(f"Suggested book for {genre} : {bookCollection}")

books(title="The Illusion of Life", author="Richard Williams", genre="animate")
```

#### positional arguments - pre-defined paramenters

```python
def positional_args(*args):

    multiplywith = 1

    for n in args:
        multiplywith *= n

    return multiplywith
    
print(f"products of all parameters : {positional_args(2, 3, 5, 7.5)}")
print(f"products of all parameters : {positional_args(5, 5, 25, 1, 1.5, 0.5)}")
```

#### factorial using recursive function

```python
def factorial(i):
    if i == 0:
        return 1
    else :
        return i * factorial(i-1)
print(factorial(5))
```

#### implementing lambda function

`n args : only one operation`

```python
oprations = lambda a, b, c, d, e, f : (a + b - c) * d // e //f
print(oprations(4, 2, 1, 4, 2, 3))
```

#### implementing map function

```python
num = {3, 6, 7, 8, 9, 0.0, 11}
print(f"sorted cube values : {sorted(set(map(lambda x:x**3, num)))}")

z= set(filter(lambda x:x%2!=0, num))

print(f"odds using filter function : {sorted(z)}")
print(f"power^4 using map function : {sorted(set(map(lambda x:x**4, z)))}")
```

#### check vote eligibility

```python
def check_voting_eligibility(age):
    if age < 0:
        return "Age cannot be negative. Please enter a valid age."
    elif age >= 18:
        return "You are eligible to vote."
    else:
        return f"You are not eligible to vote. You need {18 - age} more year(s)."
try:
    age_input = input("Enter your age :").strip()
    if not age_input.isdigit():
        raise ValueError("Invalid input. Please enter a valid positive integer.")

    age = int(age_input)
    print(f"age : {age}")
    result = check_voting_eligibility(age)
    print(result)

except ValueError as e:
    print(e)
```
