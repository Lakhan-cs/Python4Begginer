#### Find min, max and sum of element of tuple of random numbers

```python
def min_tpl(i):
    return min(i)

def max_tpl(i):
    return max(i)

def sum_tpl(i):
    return sum(i)


import random

random_tpl = tuple(random.randint(1, 100) for _ in range(20))

print(random_tpl)
print(f"Min value is {min_tpl(random_tpl)}")
print(f"Max value is {max_tpl(random_tpl)}")
print(f"Sum is {sum_tpl(random_tpl)}")
```
