#### Use random numbers and arrange in different order

```python
import random

random_num = [random.randint(1, 10) for _ in range(7)]

print("List of random elements :", random_num)

sorted_num = sorted(random_num)
print(f"Shorted list : {sorted_num}")

descending_num = sorted(random_num, reverse=True)
print(f"List elements in descending order : {descending_num}")

unique_num = list(set(random_num))
print(f"list with unique values : {unique_num}")
```

#### Find transpose of a matrice

```python
def transpose_matrix(matrix):

    transposed = [
        [matrix[j][i] for j in range(len(matrix))] 
        for i in range(len(matrix[0]))
        ]
    
    return transposed

matrix = [
    [2, 4, 6],
    [3, 6, 7],
    [30, 50, 60,]
]

transposed = transpose_matrix(matrix)

print(f"Original {len(matrix)}x{len(matrix[0])} matrix :")
for row in matrix:
    print(row)

print()

print("Transposed matrix :")
for row in transposed:
    print(row)
```

#### rotate a list n times 

```python
def rotate_lst(lst, n):
    return lst[n:] + lst[:n]

lst = [1, 2, 4, 7, 9, 33]
rotated = rotate_lst(lst, 3)

print(f"Original list : {lst}")
print(f"Rotated list : {rotated}")
```

#### Get intersection of two list

```python
def inter_lst(i, j):
    return [x for x in i if x in j]

lst1 = [2, 45, 66, 22, 3, 5, 7, 23]
lst2 = [3, 5, 24, 5, 2, 54, 74, 38]

intersection = inter_lst(lst1, lst2)

print(f"list1 : {lst1}")
print(f"list2 : {lst2}")
print(f"Intersection : {intersection}")
```
